```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Create bins
max_tenure = df['tenure'].max()
bins = np.arange(1, max_tenure + 13, 12)
labels = [f"{i}-{i+11}" for i in range(1, max_tenure+1, 12)]

# Create tenure group column
df['tenure_group'] = pd.cut(df['tenure'],
                            bins=bins,
                            labels=labels,
                            right=True,
                            include_lowest=True)

# Plot
plt.figure(figsize=(10,6))
ax = sns.countplot(data=df, x='tenure_group', hue='Churn')

plt.xlabel("Tenure (Months)")
plt.ylabel("Number of Customers")
plt.title("Customer Distribution by Tenure with Churn")
plt.xticks(rotation=0)

# Add count labels
for container in ax.containers:
    ax.bar_label(container)

plt.legend(title="Churn")
plt.tight_layout()
plt.show()
```

Though Tenure has a very low impact on Churning:
It is seen that:
customers with tenure between 1-12 mnths have churned the most (47%)
Customers with more than 55 mnths tenure have very low churning rate

### Model Training

```python
df.head()
```

```python
new_df.head()
```

```python
df.columns
```

```python
df.drop(columns=['tenure_group'],inplace=True)
```

```python
df.columns
```

```python
df=pd.get_dummies(df,columns=['gender', 'SeniorCitizen', 'Partner', 'Dependents',
       'PhoneService', 'MultipleLines', 'InternetService', 'OnlineSecurity',
       'OnlineBackup', 'DeviceProtection', 'TechSupport', 'StreamingTV',
       'StreamingMovies', 'Contract', 'PaperlessBilling', 'PaymentMethod',
       'Churn'],drop_first=True).astype(int)
```

```python
df.head()
```

```python
x=df.drop('Churn_Yes',axis=1)
```

```python
y=df.Churn_Yes
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
```

```python
#Feature Scaling
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_train_sc=sc.fit_transform(x_train)
x_test_sc=sc.transform(x_test)
```

### **Decision Tree**

```python
from sklearn.tree import DecisionTreeClassifier
dt=DecisionTreeClassifier()
dt.fit(x_train_sc,y_train)
y_pred=dt.predict(x_test_sc)
y_prob = dt.predict_proba(x_test_sc)[:,1]
threshold = 0.35
y_pred_custom = (y_prob >= threshold).astype(int)
from sklearn.metrics import accuracy_score,recall_score,roc_auc_score
print('Accuracy: ',accuracy_score(y_test,y_pred))
print('Recall: ',recall_score(y_test,y_pred_custom))
print('ROC_AUC: ',roc_auc_score(y_test,y_pred))
```

### **Random Forest**

```python
from sklearn.ensemble import RandomForestClassifier
rf=RandomForestClassifier()
rf.fit(x_train_sc,y_train)
y_pred=rf.predict(x_test_sc)
y_prob = rf.predict_proba(x_test_sc)[:,1]
threshold = 0.35
y_pred_custom = (y_prob >= threshold).astype(int)
from sklearn.metrics import accuracy_score,recall_score,roc_auc_score
print('Accuracy: ',accuracy_score(y_test,y_pred))
print('Recall: ',recall_score(y_test,y_pred_custom))
print('ROC_AUC: ',roc_auc_score(y_test,y_pred_custom))
```

### **Naive Bayes**

```python
from sklearn.naive_bayes import BernoulliNB
nb=BernoulliNB()
nb.fit(x_train_sc,y_train)
y_pred=nb.predict(x_test_sc)
y_prob = nb.predict_proba(x_test_sc)[:,1]
threshold = 0.35
y_pred_custom = (y_prob >= threshold).astype(int)
from sklearn.metrics import accuracy_score,recall_score,roc_auc_score
print('Accuracy: ',accuracy_score(y_test,y_pred))
print('Recall: ',recall_score(y_test,y_pred_custom))
print('ROC_AUC: ',roc_auc_score(y_test,y_pred_custom))
```

### **Logistic Regression**

```python
from sklearn.linear_model import LogisticRegression
lr=LogisticRegression()
lr.fit(x_train_sc,y_train)
y_pred=lr.predict(x_test_sc)
y_prob = lr.predict_proba(x_test_sc)[:,1]
threshold = 0.35
y_pred_custom = (y_prob >= threshold).astype(int)
from sklearn.metrics import accuracy_score,recall_score,roc_auc_score,precision_score
print('Accuracy: ',accuracy_score(y_test,y_pred_custom))
print('Recall: ',recall_score(y_test,y_pred_custom))
print('ROC_AUC: ',roc_auc_score(y_test,y_pred_custom))
print('Precision: ',precision_score(y_test,y_pred_custom))
```

### **Ada Boost**

```python
from sklearn.ensemble import AdaBoostClassifier
ada=AdaBoostClassifier()
ada.fit(x_train_sc,y_train)
y_pred=ada.predict(x_test_sc)
y_prob = ada.predict_proba(x_test_sc)[:,1]
threshold = 0.35
y_pred_custom = (y_prob >= threshold).astype(int)
from sklearn.metrics import accuracy_score,recall_score,roc_auc_score,precision_score
print('Accuracy: ',accuracy_score(y_test,y_pred_custom))
print('Recall: ',recall_score(y_test,y_pred_custom))
print('ROC_AUC: ',roc_auc_score(y_test,y_pred_custom))
print('Precision: ',precision_score(y_test,y_pred_custom))
```

### **XG Boost**

```python
from xgboost import XGBClassifier
xgb=XGBClassifier()
xgb.fit(x_train_sc,y_train)
y_pred=xgb.predict(x_test_sc)
y_prob = xgb.predict_proba(x_test_sc)[:,1]
threshold = 0.35
y_pred_custom = (y_prob >= threshold).astype(int)
from sklearn.metrics import accuracy_score,recall_score,roc_auc_score
print('Accuracy: ',accuracy_score(y_test,y_pred))
print('Recall: ',recall_score(y_test,y_pred_custom))
print('ROC_AUC: ',roc_auc_score(y_test,y_pred_custom))
```

Logistic Regression was better than AdaBoost because it had a significantly higher ROC-AUC (0.759 vs 0.707), meaning it ranked churners more accurately overall.

Although AdaBoost had higher recall, its lower AUC indicates weaker probability separation, which can lead to more false positives and higher 
retention costs.

Logistic Regression was selected because it provided the best balance between recall and precision, achieved the highest ROC-AUC, and reduced false positive targeting cost compared to AdaBoost.

### **HPO**

Optimizing AUC

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GridSearchCV

param_grid = {
    'C': [0.01, 0.1, 1, 10, 100],
    'penalty': ['l1', 'l2'],
    'solver': ['liblinear'],   # supports both l1 and l2
    'class_weight': [None, 'balanced']
}

lr = LogisticRegression(max_iter=1000)

grid = GridSearchCV(
    lr,
    param_grid,
    cv=5,
    scoring='roc_auc',
    n_jobs=-1
)

grid.fit(x_train_sc, y_train)

print("Best Parameters:", grid.best_params_)
print("Best CV AUC:", grid.best_score_)
```

```python
best_lr = grid.best_estimator_

y_prob = best_lr.predict_proba(x_test_sc)[:, 1]
y_pred = (y_prob >= 0.35).astype(int)

from sklearn.metrics import accuracy_score, recall_score, precision_score, roc_auc_score

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("ROC-AUC:", roc_auc_score(y_test, y_prob))
print("Precision:", precision_score(y_test, y_pred))
```

**ROC-AUC 75.7-->84.21**

**What actually is threshold:**

**The probability cutoff used to convert predicted probabilities into 0/1 class labels.**

**EG: “Target customers with ≥30% churn risk” → threshold = 0.30**

```python
import numpy as np
from sklearn.metrics import recall_score, precision_score

thresholds = np.arange(0.2, 0.6, 0.05)

for t in thresholds:
    y_pred = (y_prob >= t).astype(int)
    print("Threshold:", t,
          "Recall:", recall_score(y_test, y_pred),
          "Precision:", precision_score(y_test, y_pred))
```

After improving ROC-AUC through hyperparameter tuning, threshold optimization showed that 0.30–0.35 provides the best balance between recall and precision for churn intervention strategy.
